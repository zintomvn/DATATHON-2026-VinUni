```python
from pathlib import Path
import sys

import pandas as pd

# Find project root (folder that contains 'src')
PROJECT_ROOT = Path.cwd()
while PROJECT_ROOT != PROJECT_ROOT.parent and not (PROJECT_ROOT / 'src').exists():
    PROJECT_ROOT = PROJECT_ROOT.parent
sys.path.append(str(PROJECT_ROOT))

from src.data.load_data import load_all_raw, load_sales

raw = load_all_raw(PROJECT_ROOT / 'data' / 'raw')
{k: v.shape for k, v in raw.items()}
```

```python
from src.data.preprocess import clean_sales, ensure_daily_continuity

sales = clean_sales(load_sales(PROJECT_ROOT / 'data' / 'raw'))
ensure_daily_continuity(sales)
sales.head()
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(12, 4))
plt.plot(sales['Date'], sales['Revenue'])
plt.title('Daily Revenue (Train)')
plt.xlabel('Date')
plt.ylabel('Revenue')
plt.tight_layout()
plt.show()
```

```python
# Example: correlate web traffic with revenue (overlapping period only)
web = raw['web_traffic'].copy()
web['date'] = pd.to_datetime(web['date'], errors='coerce')
daily_web = (
    web.dropna(subset=['date'])
       .groupby('date', as_index=False)
       .agg({
            'sessions': 'sum',
            'unique_visitors': 'sum',
            'page_views': 'sum',
            'bounce_rate': 'mean',
            'avg_session_duration_sec': 'mean',
        })
)
merged = sales.merge(daily_web, left_on='Date', right_on='date', how='inner')
merged[['Revenue', 'sessions', 'unique_visitors', 'page_views']].corr()
```
